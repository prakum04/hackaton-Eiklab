```python
#removing columns with all missing values
df = df_org.copy()
df = df.dropna(axis=1, how="all")
```

```python
#checking missing values for columns with 100% association
a = "mgmt_rec_raw"
b = "mgmt_rec" 

# Count each combination of missing / not missing
print(pd.crosstab(
    df[a].isna(),
    df[b].isna(),
    rownames=[f"{a} missing"],
    colnames=[f"{b} missing"]
))

# Show rows where only one of the two columns is missing
mismatch = df[a].isna() != df[b].isna()

print("Rows with mismatched missing values:", mismatch.sum())
display(df.loc[mismatch, [a, b]])
```

```python
a = "nbim_vote_raw"
b = "nbim_vote" 

# Count each combination of missing / not missing
print(pd.crosstab(
    df[a].isna(),
    df[b].isna(),
    rownames=[f"{a} missing"],
    colnames=[f"{b} missing"]
))

# Show rows where only one of the two columns is missing
mismatch = df[a].isna() != df[b].isna()

print("Rows with mismatched missing values:", mismatch.sum())
display(df.loc[mismatch, [a, b]])
```
